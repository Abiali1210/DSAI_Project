### Raw Dataset Dimensions & Structure
The dataset has 862,927 rows and 33 columns (features + target).

```python
# Data types and non-null counts
print("Column info:")
df.info()
```

- All 33 columns have 862,927 non-null entries except weather columns (`PRCP`, `TMAX`, `TMIN`, `AWND`, `GustSpeed2`), which have slightly fewer (~837,000). This is mostly weather data, which means roughly 3% of weather data is missing.

- Data types:
  - `object`: Date/time columns, zip codes, car category, color, make, model. These will need encoding later.
  - `float64`: Numeric features like distance, location, ratings, weather.
  - `int64`: Binary & integer features like `cancelled`, `Holiday`, `Weekend`.

So missing weather data will need imputation before modeling.

Categorical columns will need to be converted to numeric format through encoding.

The target `cancelled` is already integer (0/1), which is ready for classification.

```python
# Summary statistics
print("Summary statistics:")
df.describe()
```

- `Distance Travelled`: Min value is -1.0, which is not physically possible. So `-1` is used as a placeholder for missing data.
- `Driver Rating`: Min is -1.0, median is 5.0. Again, `-1` likely means it is missing or no rating recorded.
- `Surge Factor`: Mostly 1.0 (no surge), max 6.0. High surge entries are rare.
- Weather columns (`PRCP`, `TMAX`, etc.): Show some extreme values (e.g., TMAX min = -67.8) that seem unrealistic. These may be data errors.
- `cancelled`: Mean = 0.0125 → only 1.25% of rides are cancelled. This is a highly imbalanced target.

The low cancellation rate means we must use appropriate metrics (precision, recall, F1) and possibly handle class imbalance during modeling.

```python
# Count missing values per column
print(df.isnull().sum())
```

Interpretation:
Most columns are complete as they have 0 missing values.
Weather columns have significant missing data (roughly 25.6k each in PRCP, TMAX, TMIN, AWND, GustSpeed2).
This suggests weather data may only be available for some of the rides/dates, or was collected inconsistently.
Since these are numeric features, we can consider either: 
- imputation (median/mean)
- creating a "weather_missing" indicator flag
- dropping them if they add little predictive value
