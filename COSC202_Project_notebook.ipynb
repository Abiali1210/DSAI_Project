**Class Imbalance Implications:**

Only **1.25%** of rides are cancelled (10,777 out of 862,927). This is a **severe class imbalance**.

- **Accuracy trap**: A model that predicts "not cancelled" for every ride would achieve ~98.75% accuracy while being completely useless — it would miss every cancellation.
- **Metric selection**: We must rely on Precision, Recall, F1-score, and ROC-AUC rather than accuracy. Recall (catching actual cancellations) is particularly important from a business standpoint.
- **Modeling strategy**: Will need class weighting (`class_weight='balanced'`), SMOTE, or other resampling techniques during Task 2 to help the model learn the minority class patterns.

Note: This analysis is for understanding only. Final decisions on handling imbalance will be made in Task 2.

```python
# Outlier detection using IQR method and unrealistic value checks
# NOTE: These statistics are for data understanding only - not for making final decisions on the dataset

import numpy as np

print("=== IQR-based Outlier Detection ===\n")
numeric_cols = df.select_dtypes(include=[np.number]).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    pct = (outliers / len(df)) * 100
    if outliers > 0:
        print(f"{col}: {outliers:,} outliers ({pct:.1f}%) | Range: [{df[col].min()}, {df[col].max()}]")

print("\n=== Unrealistic Value Flags ===")
print(f"TMAX < 0: {(df['TMAX'] < 0).sum()} rows (impossible for max temp in F for this region)")
print(f"TMIN < 0: {(df['TMIN'] < 0).sum()} rows (possible but extreme)")
print(f"TMIN > TMAX: {(df['TMIN'] > df['TMAX']).sum()} rows (min temp cannot exceed max temp)")
print(f"Year > 2026: {(df['Year'] > 2026).sum()} rows (future manufacturing year)")
print(f"Distance > 1,000,000 m: {(df['Distance Travelled'] > 1000000).sum()} rows (unrealistic distance)")
```

**Outlier Observations:**

- **Weather extreme values**: TMAX < 0 in 4,001 rows and TMIN < 0 in 4,390 rows — physically implausible for this region (Austin, TX, ~30°N). Additionally, **TMIN > TMAX in 8,235 rows**, which is a logical impossibility confirming these are data entry or sensor errors.

- **Year > 2026**: 242 rows have future manufacturing years (e.g., 2104). These are likely typos (2014 → 2104).

- **Distance > 1,000,000 m**: Only 16 rows exceed 1,000 km — clearly erroneous. However, the max value of 43M meters (~43,000 km) is extreme.

- **IQR outliers**: 20.3% of Driver Rating values are flagged as outliers — this is largely due to the `-1` placeholder values pulling the lower bound. Once `-1` is handled, this will normalize. Similarly, Surge Factor (13.8%) and binary columns (Holiday, Thunder) show IQR outliers because the IQR method isn't designed for skewed or binary distributions.

Note: These issues will be addressed during preprocessing in Task 2.

```python
# Correlation analysis to identify redundant features
# NOTE: This is for data understanding only - not for making final decisions on the dataset

# Select numeric columns for correlation (exclude location coords to focus on meaningful features)
numeric_for_corr = df[['Distance Travelled', 'Driver Rating', 'Rider Rating', 'Surge Factor',
                        'Rating', 'PRCP', 'TMAX', 'TMIN', 'AWND', 'GustSpeed2',
                        'Holiday', 'Weekend', 'cancelled', 'Number of rides that day']].copy()

corr_matrix = numeric_for_corr.corr()

# Show highly correlated pairs (|r| > 0.7)
print("Highly correlated feature pairs (|r| > 0.7):")
high_corr = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        val = corr_matrix.iloc[i, j]
        if abs(val) > 0.7:
            high_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], val))

if high_corr:
    for f1, f2, val in high_corr:
        print(f"  {f1} <-> {f2}: r = {val:.4f}")
else:
    print("  No pairs with |r| > 0.7")

# Show correlations with the target
print("\nCorrelation with 'cancelled':")
target_corr = corr_matrix['cancelled'].sort_values(key=abs, ascending=False)
for feat, val in target_corr.items():
    if feat != 'cancelled':
        print(f"  {feat}: {val:.4f}")
```

**Correlation & Redundancy Findings:**

- **No highly correlated pairs (|r| > 0.7)**: Even TMAX and TMIN don't exceed 0.7 linear correlation. This suggests that while these features are related, they capture distinct aspects of temperature variation.

- **Extremely weak linear correlation with target**: The strongest correlation with `cancelled` is Rider Rating (r = -0.11), followed by Driver Rating (r = -0.05) and Rating (r = -0.03). All other features have |r| < 0.02. **This does NOT mean these features are unimportant** — cancellations are likely driven by **non-linear patterns and feature interactions** (e.g., certain combinations of weather + time + driver state) that Pearson correlation cannot detect. Tree-based models should be able to capture these.

- **Negative direction**: Both Driver Rating and Rider Rating are negatively correlated with cancellation — lower ratings are associated with higher cancellation rates. This aligns with intuition.

### Missing Data Hypotheses

**Weather columns** (`PRCP`, `TMAX`, `TMIN`, `AWND`, `GustSpeed2`) — ~25,600 missing each (~3%):
- **Type**: Likely **MAR** (Missing At Random) — the missingness depends on date/location. External weather station data may not cover all ride dates or areas.
- **Hypothesis**: Weather data was merged from an external API or station and certain date ranges or geographic zones lacked matching records.
- **Suggested approach**: Median imputation, possibly grouped by month/season. Optionally add a "weather_missing" indicator flag.

**-1 values** (Driver Rating, Rider Rating, Zip Codes, Distance):
- **Type**: **Informative missingness** (a form of MNAR — Missing Not At Random). The missingness itself carries information.
- **Hypothesis**: Driver Rating = -1 means the driver was new at booking time (11.65% of rides). Distance = -1 is *caused by* the cancellation (ride never started). Zip Code = -1 means GPS was used instead of zip lookup.
- **Suggested approach**: For ratings, impute with median and add a "no_prior_rating" indicator. For distance, preserve -1 as a meaningful category or create a flag.

Note: These strategies will be implemented in Task 2 (Cleaning & Preprocessing).

```python
# Data Quality Summary Table
# NOTE: This is for data understanding only - not for making final decisions on the dataset

summary_data = []
for col in df.columns:
    null_count = df[col].isna().sum()
    minus_one = 0
    if df[col].dtype in ['int64', 'float64']:
        minus_one = (df[col] == -1).sum()
    elif df[col].dtype == 'object':
        minus_one = (df[col] == '-1').sum()

    issues = []
    if null_count > 0:
        issues.append(f"{null_count:,} NaN")
    if minus_one > 0:
        issues.append(f"{minus_one:,} as -1")
    if df[col].dtype in ['int64', 'float64']:
        if df[col].min() < 0 and 'Long' not in col and col not in ['Start Location Long', 'End Location Long']:
            issues.append(f"negative min={df[col].min()}")

    summary_data.append({
        'Column': col,
        'Type': str(df[col].dtype),
        'Nulls': f"{null_count:,}" if null_count > 0 else "0",
        'Issues': '; '.join(issues) if issues else 'Clean'
    })

summary_df = pd.DataFrame(summary_data)
pd.set_option('display.max_colwidth', 60)
print(summary_df.to_string(index=False))
```
